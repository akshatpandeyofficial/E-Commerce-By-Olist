```python
datasets = { "customers" : customers,
             "geolocation" : geolocation,
             "orderitem" : orderitem,
             "payments" : payments,
             "reviews" : reviews,
             "orders" : orders,
             "products" : products,
             "sellers" : sellers,
             "Category_translation" : Category_translation }


for name , df in datasets.items():
    print("--------------------------------------")
    print("DATASET: ", name)
    print("Shape: ", df.shape)
    print("Columns: ", df.columns.tolist())
    print("Missing Values: ", df.isna().sum().sum())
    print("Duplicated Values: ", df.duplicated().sum())
    print("--------------------------------------")
```

# Identifying Missing Values In the Datasets

```python
for Name, df in datasets.items():
    print("Name: " , Name)
    missing = df.isnull().sum()
    missing = missing[missing>0]
    print(missing.sort_values(ascending=False))
    print("-----------------------------")
```

# Identfying Duplicated Records in Datasets


```python
for name , df in datasets.items():
    print(name,":", df.duplicated().sum())
```

# Data Cleaning -:

```python
#Creating New Dictonary for storing the cleaned Value in there

Cleaned_dataset = {}
for name , df in datasets.items():
    Cleaned_dataset[name]= df.copy()

print("New Cleaned_Dataset Dictonary Created! ")
for name ,df in Cleaned_dataset.items():
    print(name, df.shape)
```

```python
#Removing exact Duplicate Records 
```

```python
for name , df in Cleaned_dataset.items():
    Cleaned_dataset[name]=df.drop_duplicates()
print("Duplicate Rows Removed")
```

```python
#Checking Wether Duplicated Rows Were Removed Or Not
```

```python
for name, df in Cleaned_dataset.items():
    print(name, ":", df.duplicated().sum())
```

```python
#Cleaning The Table Reviews Where Review is not fill is Replaced by "NO COMMENT"
```

```python
Cleaned_dataset["reviews"]["review_comment_title"]= (Cleaned_dataset["reviews"]["review_comment_title"].fillna("No Comment"))

Cleaned_dataset["reviews"]["review_comment_message"]=(Cleaned_dataset["reviews"]["review_comment_message"].fillna("No Comments"))

print("Review Cleaned Sucessfully!")
```
